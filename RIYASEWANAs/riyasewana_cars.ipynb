# Cars Category

```python
from selenium import webdriver
from selenium.webdriver.common.keys import Keys
import time
# from bs4 import BeautifulSoup
# import requests
import pandas as pd
```

```python
driver = webdriver.Chrome(executable_path='C:\selenium_browser_drivers\chromedriver.exe')

locations = []
add_dates = []
prices = []
makes = []
models = []
model_years = []
gears = []
fuel_types = []
engine_ccs = []
mileage_km = []

for num in range(1,345):
    if num == 1:
        url = 'https://riyasewana.com/search/cars'
    else:
        url = 'https://riyasewana.com/search/cars?page='+str(num)
    
    print('Started Page no.'+str(num)+'. - ', url)
    
    if num == 344:
        print("PLEASE NOTE - You have reached the last page. If I'm gonna throw an error on this page, ignore me and run the rest of the code as last page may not contain 60 elements. If not, don't worry about what I said..")
    
    time.sleep(2)
    driver.get(url)
    
    driver.implicitly_wait(5)
    
    for i in range(1,61):
        path = "//*[@id='content']/ul/li["+ str(i)+ "]/h2/a"
        driver.find_element_by_xpath(path).click()
#         time.sleep(2)
        try:
            location = driver.find_element_by_xpath("//*[@id='content']/table/tbody/tr[4]/td[2]").text
            locations.append(location)
     
            add_date = driver.find_element_by_xpath("//*[@id='content']/table/tbody/tr[4]/td[4]/time").text
            add_dates.append(add_date)

            price = driver.find_element_by_xpath("//*[@id='content']/table/tbody/tr[5]/td[2]").text
            prices.append(price)

            make = driver.find_element_by_xpath("//*[@id='content']/table/tbody/tr[5]/td[4]").text
            makes.append(make)

            model = driver.find_element_by_xpath("//*[@id='content']/table/tbody/tr[6]/td[2]").text
            models.append(model)

            model_year = driver.find_element_by_xpath("//*[@id='content']/table/tbody/tr[6]/td[4]").text
            model_years.append(model_year)

            gear = driver.find_element_by_xpath("//*[@id='content']/table/tbody/tr[7]/td[2]").text
            gears.append(gear)

            fuel_type = driver.find_element_by_xpath("//*[@id='content']/table/tbody/tr[7]/td[4]").text
            fuel_types.append(fuel_type)

            engine_cc = driver.find_element_by_xpath("//*[@id='content']/table/tbody/tr[8]/td[4]").text
            engine_ccs.append(engine_cc)

            mileage = driver.find_element_by_xpath("//*[@id='content']/table/tbody/tr[9]/td[4]").text
            mileage_km.append(mileage)
            
        except Exception as e:
            print(e)
        driver.back()
#         time.sleep(1)
```

```python
df = pd.DataFrame({'location':locations, 'add_date':add_dates, 'price':prices, 'make':makes, 'model':models,
                  'model_year':model_years, 'gear':gears, 'fuel_type':fuel_types, 
                   'engine_cc':engine_ccs, 'mileage_km':mileage_km})
```
