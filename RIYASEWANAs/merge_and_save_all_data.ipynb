```python
import pandas as pd
# import numpy as np
from functools import reduce
```

```python
cars = pd.read_csv('cars.csv')
buses = pd.read_csv('buses.csv')
crew_cabs = pd.read_csv('crew_cabs.csv')
heavy = pd.read_csv('heavy.csv')
lorries = pd.read_csv('lorries.csv')
pickup = pd.read_csv('pickup.csv')
suvs = pd.read_csv('suv.csv')
tractors = pd.read_csv('tractor.csv')
trucks = pd.read_csv('trucks.csv')
tuks = pd.read_csv('tuk.csv')
vans = pd.read_csv('vans.csv')
wagons = pd.read_csv('wagon.csv')
```

```python
data_frames = [cars, buses, crew_cabs, heavy, lorries, pickup,suvs, tractors, trucks, tuks, vans, wagons]
```

```python
merged_df = pd.concat(data_frames)
```
